## Juggling between NumPy and pandas
-  The pandas library is actually built on top of NumPy. However, you do need to make a choice between NumPy and pandas when performing tasks. You need the low‐level functionality of NumPy to perform some tasks, but pandas makes things so much easier that you want to use it as often as possible. 

### Knowing when to use NumPy
- It’s essential to realize that developers built pandas on top of NumPy. As a result, every task you perform using pandas also goes through NumPy. To obtain the benefits of pandas, you pay a performance penalty that some testers say is 100 times slower than NumPy for a similar task.
- So when speed is essential, NumPy is always the better choice.

### Knowing when to use pandas
- You use pandas to make writing code easier and faster. Because pandas does a lot of the work for you. 
- The essential consideration, though, is that the pandas library provides rich time‐series functionality, data alignment, NA‐friendly statistics, groupby, merge, and join methods. Normally, you need to code these features when using NumPy, which means
you keep reinventing the wheel.

### Why NumPy? 
- One of the biggest advantages of using the NumPy package is the ndarray (n-dimensional array) data structure. The NumPy ndarray is much more powerful than the python list, and provides a larger variety of operations and functions than a python array .

### Array Manipulation
- The ndarray object allows us to perform arithmetic operations element-wise on two arrays of the same size:

```python
a=np.array([56,120,80,450])
b=np.array([20,5,6,211])
print(a-b)
print(a+b)
print(a*b) # * is for element wise multiplication
```

```python
a=np.array([[2,3,4],[2,3,4]])
b=np.array([[2,3],[3,4],[4,5]])
print(a@b) # @ is for matrix multiplication
```

### Numpy provides various <font color="blue">aggregation functions</font> to perform operations on arrays. Aggregation functions in NumPy allow you to perform computations across the entire array or along a specified axis. Here are some commonly used NumPy aggregation functions:
- min/max
- sum
- mean
- median

```python
a= np.array([[1,5,6],
           [4,7,0],
            [2,9,3]])
print(a)
```

```python
print(a*2)
```

```python
print(a.min(axis=1))
print(a.argmin(axis=1)) #Returns the indices of the minimum values along an axis.
print(a.argmin(axis=0))
```

```python
print(a.max(axis=0))
print(a.argmax(axis=0)) #Returns the indices of the minimum values along an axis.
```

```python
print(a.sum())
print(a.sum(axis=1))
print(a.sum(axis=0))
```

```python
print(a.mean())
print(a.mean(axis=0))
print(a.mean(axis=1))
```

### Indexing and Slicing

#### Integer Indexing

```python
a=np.arange(10)
print(a)
print(a[0])
print(a[:4])
```

```python
a=np.array([(1,2,3,4),(3,4,5,6),(5,6,7,8)])
print("1==",a)
print("2==",a[0][2])
print("3==",a[:1])
print("4==",a[0:2])
print("5==",a[1,3])
print("6==",a[:2,:2])
print("7==",a[0:2,[1,2]])
print("8==",a[:,::-1])
```

#### Boolean Indexing

```python
a=np.arange(10)
print(a[a>5]) 
```

```python
a = np.array([np.nan, 1,2,np.nan,3,4,5]) 
print(a[~np.isnan(a)])
```

```python
a = np.array([1, 2+6j, 5, 3.5+5j]) 
print(a[np.iscomplex(a)])
```

### Get Elements Based on a Condition--np.where() is used
    np.where(cond[, x, y]) — if cond condition is satisfied, return x , else return y .

```python
# np.where( ) returns a 1 at the indices that hold values greater than 25, and 0 where they are 25 or less.
a=np.array([(1,2,3,4),(3,4,5,6),(5,6,7,8)])
np.where(a>5,1,0)
```

```python
# If you wanted to find low values (e.g. < 15) and replace them with a -1:
np.where(a<5,-1,a)
```

### numpy.insert()

```python
a = np.array([[1,2],[3,4],[5,6]]) 
#'Axis parameter not passed. The input array is flattened before insertion.'
print(np.insert(a,3,[11,12]) )
# Axis parameter passed. The values array is broadcast to match input array.'
print(np.insert(a,1,[11],axis = 0))
# Broadcast along axis 1
print(np.insert(a,1,11,axis = 1))
```

### Basic usage of np.delete()
- np.delete() takes three parameters as follows.
    - numpy.delete(arr, obj, axis=None)
    - arr: Input array
    - obj: Row or column number to delete
    - "axis: Axis to delete

```python
a = np.arange(12).reshape(3,4) 
print(a)
#'Array flattened before delete operation as axis not used:' 
print(np.delete(a,4))
# dele column two
print(np.delete(a,1,axis = 1)) 
```

### Numpy ufunc | Universal functions
- These functions include standard trigonometric functions, functions for arithmetic operations, handling complex numbers, statistical functions, etc. Universal functions have various characteristics which are as follows-
    - It performs fast element-wise array operations.
    - It supports various features like array broadcasting, type casting etc.
    - Numpy, universal functions are objects those belongs to numpy.ufunc class.
    - Python functions can also be created as a universal function using frompyfunc library function.
    - Some ufuncs are called automatically when the corresponding arithmetic operator is used on arrays. For example when addition of two array is performed element-wise using ‘+’ operator then np.add() is called internally.

```python
# trignometric function 
angles = np.array([0, 30, 45, 60, 90, 180])
radians = np.deg2rad(angles) 
print(np.sin(radians)) 
print(np.sinh(radians)) 
```

```python
# Statistical functions:
weight = np.array([50.7, 52.5, 50, 58, 55.63, 73.25, 49.5, 45]) 
print(np.amin(weight), np.amax(weight)) 
print(np.percentile(weight, 70)) 
print(np.mean(weight))
print(np.average(weight)) 
```

```python
# Math functions
x1 = np.arange(9.0).reshape((3, 3))
x2 = np.arange(3.0)
print(np.add(x1,x2))
print(np.subtract(x1, x2))
print(np.power(x1, 3))
```
