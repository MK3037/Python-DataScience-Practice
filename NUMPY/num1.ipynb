#### Comparison of plain Python lists vs NumPy arrays for element-wise operations.

##### 1. Element-wise Addition: List vs NumPy

```python
import time
import numpy as np
```

```python
# Plain Python list addition using zip
a = [1, 2, 3, 4, 5]
b = [1, 2, 3, 4, 5]

start = time.time()
c = [a + b for a, b in zip(a, b)]
end = time.time()

print('Result:', c)
print(f'Time taken: {end - start:.8f} seconds')
```

```python
# NumPy array addition — faster and cleaner
x = np.array([1, 2, 3, 4, 5])
y = np.array([1, 2, 3, 4, 5])

start = time.time()
z = x + y
end = time.time()

print('Result:', z)
print(f'Time taken: {end - start:.8f} seconds')
```

##### 2. Squaring Elements: List vs NumPy

```python
# List comprehension needed for element-wise square
print('List:', [x * x for x in a])

# NumPy applies the operation directly — no loop needed
print('NumPy:', x ** 2)
print('NumPy:', x+10)       #broadcasting

arr1 = np.array([1, 2, 3])
arr2 = np.array([10, 20, 30])
result = arr1 + arr2 
print(result)  

arr1 = np.array([[1, 2, 3],[4 ,5 ,6]])
result = arr1 + arr2  
print(result) 
```

##### 3. Array Creation Utilities

```python
print('zeros:   ', np.zeros((2, 3)))
print('ones:    ', np.ones((2, 2)))
print('full:    ', np.full((2, 3), 7))
print('arange (step=2):', np.arange(1, 10, 2))
print('arange (0-2):   ', np.arange(3))
print('linspace:       ', np.linspace(0, 10, num=5))
```
