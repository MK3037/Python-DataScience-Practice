# 05 · Sorting Data

Pandas' `sort_values` and `sort_index` are the equivalent of SQL's `ORDER BY`.

```python
import pandas as pd

people = {
    'first': ['Corey', 'Jane', 'John'],
    'last': ['Schafer', 'Doe', 'Doe'],
    'email': ['CoreyMSchafer@gmail.com', 'JaneDoe@email.com', 'JohnDoe@email.com']
}
df = pd.DataFrame(people)
# SQL: CREATE TABLE people (first TEXT, last TEXT, email TEXT); INSERT INTO people VALUES (...);
```

```python
df.sort_values(by='last', ascending=False)
# SQL: SELECT * FROM people ORDER BY last DESC;
```

```python
df.sort_values(by=['last', 'first'], ascending=False, inplace=True)
df
# SQL: SELECT * FROM people ORDER BY last DESC, first DESC;
```

```python
df.sort_values(by=['last', 'first'], ascending=[False, True])
# SQL: SELECT * FROM people ORDER BY last DESC, first ASC;
```

```python
df.sort_index()
# SQL: SELECT * FROM people ORDER BY id ASC;   -- id = the row's original index / primary key
```

```python
df['last'].sort_values()
# SQL: SELECT last FROM people ORDER BY last ASC;
```
