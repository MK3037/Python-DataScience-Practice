# 07 · Adding & Removing Rows/Columns

Pandas' equivalents of `ALTER TABLE ... ADD/DROP COLUMN`, `INSERT INTO`, and `DELETE FROM`.

```python
import pandas as pd

people = {
    'first': ['Corey', 'Jane', 'John'],
    'last': ['Schafer', 'Doe', 'Doe'],
    'email': ['CoreyMSchafer@gmail.com', 'JaneDoe@email.com', 'JohnDoe@email.com']
}
df = pd.DataFrame(people)
```

```python
df['fullname'] = df['first'] + " " + df['last']
df
# SQL: SELECT *, CONCAT(first, ' ', last) AS fullname FROM people;
# to persist it as a real column:
# ALTER TABLE people ADD COLUMN fullname TEXT;
# UPDATE people SET fullname = CONCAT(first, ' ', last);
```

```python
df.drop(columns=['first', 'last'], inplace=True)
df
# SQL: ALTER TABLE people DROP COLUMN first;
#      ALTER TABLE people DROP COLUMN last;
```

```python
df['fullname'].str.split(' ', expand=True)   # preview — splits into separate columns, not saved yet
# SQL: SELECT SUBSTRING_INDEX(fullname, ' ', 1) AS first,
#             SUBSTRING_INDEX(fullname, ' ', -1) AS last
#      FROM people;
```

```python
df[['first', 'last']] = df['fullname'].str.split(' ', expand=True)
df
# SQL: ALTER TABLE people ADD COLUMN first TEXT, ADD COLUMN last TEXT;
#      UPDATE people SET first = SUBSTRING_INDEX(fullname, ' ', 1),
#                         last  = SUBSTRING_INDEX(fullname, ' ', -1);
```

```python
df = pd.concat([df, pd.DataFrame([{'first': 'Tony'}])], ignore_index=True)
df
# SQL: INSERT INTO people (first) VALUES ('Tony');
```

```python
df.drop(index=3, inplace=True)
df
# SQL: DELETE FROM people WHERE id = 3;
```

```python
filt = df['last'] == 'Schafer'
df.drop(index=df[filt].index, inplace=True)
df
# SQL: DELETE FROM people WHERE last = 'Schafer';
```

```python
data = {
  'street': [
    '5230 Newell Road',
    '1381 9th Avenue',
    '1709 Broderick St',
    '2640 Steiner St',
    '2601 Lyon St',
    '1600 Amphitheatre Parkway',
    '7500 N Glenoaks Boulevard',
    '4501 Glencoe Avenue',
    '232 Santa Margarita Ave',
    '471 Emerson Street'
  ],
  'city': [
    'Palo Alto',
    'San Francisco',
    'San Francisco',
    'San Francisco',
    'San Francisco',
    'Mountain View',
    'Burbank',
    'Marina del Rey',
    'Menlo Park',
    None
  ],
  'state': [
    'CA',
    'CA',
    'CA',
    'CA',
    'CA',
    'CA',
    'CA',
    'CA',
    None,
    None
  ],
  'zip_code': [
    94303,
    94122,
    94115,
    94123,
    94123,
    94043,
    91504,
    90292,
    94025,
    None
  ],
  'delivery_type': [
    'First Class',
    'Standard',
    'Priority',
    'Standard',
    None,
    'Expresss',
    'Standard',
    'Standard',
    None,
    None
  ]
}

df = pd.DataFrame(data)
```

```python
cleaned_df = df.dropna()
cleaned_df
```

```python
# if we only care about missing values in specific columns
cleaned_df = df.dropna(subset = ['state', 'city'])
cleaned_df
```

```python
# The following code would fill in the missing values for the location column
df['delivery_type'] = df['delivery_type'].fillna('Standard')    #4
df
```

```python
# Finally, assuming the data is numerical, we can fill empty values with the average using .mean().
mean_zip_code = df['zip_code'].mean()

df['zip_code'] = df['zip_code'].fillna(mean_zip_code)
```
