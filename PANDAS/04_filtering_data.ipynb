# 04 · Filtering Data

This is pandas' version of SQL's `WHERE` clause. You build a **boolean mask**
(a Series of True/False, one per row) and use it to keep only matching rows.

```python
import pandas as pd

people = {
    'first': ['Corey', 'Jane', 'John'],
    'last': ['Schafer', 'Doe', 'Doe'],
    'email': ['CoreyMSchafer@gmail.com', 'JaneDoe@email.com', 'JohnDoe@email.com']
}
df = pd.DataFrame(people)
# SQL: CREATE TABLE people (first TEXT, last TEXT, email TEXT);
#      INSERT INTO people VALUES (...three rows above...);
```

```python
df['last'] == 'Doe'
# Produces one True/False per row — like a CASE expression, not a filter yet.
# SQL: SELECT last = 'Doe' AS is_doe FROM people;
```

```python
filt = (df['last'] == 'Doe')            # filt is just a variable holding the boolean mask
df[filt]                                  # df[df['last']=='Doe'] is identical, and so is df.loc[filt]
# SQL: SELECT * FROM people WHERE last = 'Doe';
```

```python
filt = (df['last'] == 'Doe') & (df['first'] == 'John')     # & = AND
df.loc[filt, ['email']]
# SQL: SELECT email FROM people WHERE last = 'Doe' AND first = 'John';
```

```python
filt = (df['last'] == 'Doe') | (df['first'] == 'John')     # | = OR
df.loc[filt, ['email']]
# SQL: SELECT email FROM people WHERE last = 'Doe' OR first = 'John';
```

```python
df.loc[~filt, ['email']]     # ~ negates the filter (NOT)
# SQL: SELECT email FROM people WHERE NOT (last = 'Doe' OR first = 'John');
```

---
### The same ideas on the real survey dataset

```python
highsalary = (df['ConvertedComp'] > 70000)
df.loc[highsalary, ['Country', 'LanguageWorkedWith', 'ConvertedComp']]
# SQL: SELECT Country, LanguageWorkedWith, ConvertedComp
#      FROM survey_results_public
#      WHERE ConvertedComp > 70000;
```

```python
countries = ['United States', 'India', 'Germany']
filt = df['Country'].isin(countries)
df.loc[filt]
# SQL: SELECT * FROM survey_results_public WHERE Country IN ('United States','India','Germany');
```

```python
df['LanguageWorkedWith']   # it's a string column, tricky to filter on directly (multi-value text)
# SQL: SELECT LanguageWorkedWith FROM survey_results_public;
```

```python
filt = df['LanguageWorkedWith'].str.contains('Python', na=False)   # na=False drops NaN rows instead of erroring
df.loc[filt, ['LanguageWorkedWith']]
# SQL: SELECT LanguageWorkedWith FROM survey_results_public WHERE LanguageWorkedWith LIKE '%Python%';
```
