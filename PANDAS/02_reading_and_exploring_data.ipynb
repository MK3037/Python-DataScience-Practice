# 02 · Reading & Exploring Data

Loading a CSV and getting a first look at it — the pandas version of
connecting to a database and running some quick `DESCRIBE` / `LIMIT` queries.

```python
import os
import pandas as pd

os.chdir(r"C:\Users\purve\OneDrive\Desktop\extra\2019survey")
# (just changes the working directory so the CSV files below can be found by name — no SQL equivalent)
```

```python
df = pd.read_csv('survey_results_public.csv')
schema_df = pd.read_csv('survey_results_schema.csv')
# SQL: this is like importing a CSV into a table, e.g. (MySQL):
# LOAD DATA INFILE 'survey_results_public.csv'
# INTO TABLE survey_results_public
# FIELDS TERMINATED BY ',' ENCLOSED BY '"'
# LINES TERMINATED BY '\n'
# IGNORE 1 ROWS;
```

```python
df.info()
# SQL: DESCRIBE survey_results_public;
#      -- (equivalent: SHOW COLUMNS FROM survey_results_public;)
```

```python
df.head()   # first 5 rows
# SQL: SELECT * FROM survey_results_public LIMIT 5;
```

```python
df.tail(10)   # last 10 rows
# SQL: SELECT * FROM survey_results_public ORDER BY Respondent DESC LIMIT 10;
# (SQL tables have no inherent order, so "last rows" only makes sense once you ORDER BY something,
#  e.g. the primary key)
```

```python
df.shape   # (rows, columns)
# SQL: SELECT COUNT(*) FROM survey_results_public;                                  -- row count
#      SELECT COUNT(*) FROM information_schema.columns WHERE table_name='survey_results_public'; -- column count
```
