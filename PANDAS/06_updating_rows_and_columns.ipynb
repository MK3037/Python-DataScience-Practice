# 06 · Updating Rows & Columns

Renaming columns, editing cell values, and applying functions —
pandas' equivalent of SQL's `ALTER TABLE ... RENAME COLUMN` and `UPDATE ... SET`.

```python
import pandas as pd

people = {
    'first': ['Corey', 'Jane', 'John'],
    'last': ['Schafer', 'Doe', 'Doe'],
    'email': ['CoreyMSchafer@gmail.com', 'JaneDoe@email.com', 'JohnDoe@email.com']
}
df = pd.DataFrame(people)
```

```python
df.columns   # UPDATING COLUMNS
# SQL: SELECT column_name FROM information_schema.columns WHERE table_name = 'people';
```

```python
df.columns = ['first_name', 'last_name', 'email']
df.columns
# SQL: ALTER TABLE people RENAME COLUMN first TO first_name;
#      ALTER TABLE people RENAME COLUMN last TO last_name;
```

```python
df.columns = [x.upper() for x in df.columns]
df
# SQL: ALTER TABLE people RENAME COLUMN first_name TO FIRST_NAME;
#      ALTER TABLE people RENAME COLUMN last_name TO LAST_NAME;
#      ALTER TABLE people RENAME COLUMN email TO EMAIL;
```

```python
df.columns = df.columns.str.replace('_', ' ')
df
# SQL: ALTER TABLE people RENAME COLUMN FIRST_NAME TO `FIRST NAME`;
#      ALTER TABLE people RENAME COLUMN LAST_NAME TO `LAST NAME`;
```

```python
df.rename(columns={'FIRST NAME': 'first_name', 'LAST NAME': 'last_name'}, inplace=True)
df
# SQL: ALTER TABLE people RENAME COLUMN `FIRST NAME` TO first_name;
#      ALTER TABLE people RENAME COLUMN `LAST NAME` TO last_name;
```

```python
df.loc[[0], ['first_name', 'last_name']] = ['Mihir', 'Kurani']     # UPDATING DATA
df.loc[[0], ['EMAIL']] = 'NA'
df
# SQL: UPDATE people SET first_name = 'Mihir', last_name = 'Kurani' WHERE id = 0;
#      UPDATE people SET EMAIL = 'NA' WHERE id = 0;
```

```python
df.loc[0] = ['Corey', 'Schafer', 'CoreyMSchafer@gmail.com']
df
# SQL: UPDATE people SET first_name = 'Corey', last_name = 'Schafer', EMAIL = 'CoreyMSchafer@gmail.com'
#      WHERE id = 0;
```

```python
# filt = (df['email'] == 'JohnDoe@email.com')
# df[filt]['last'] = 'Smith'      # ERROR — you can READ data through a filter, but not WRITE like this

filt = (df['EMAIL'] == 'JohnDoe@email.com')      # this works instead:
df.loc[filt, 'last_name'] = 'Smith'
df
# SQL: UPDATE people SET last_name = 'Smith' WHERE EMAIL = 'JohnDoe@email.com';
```

```python
df['EMAIL'] = df['EMAIL'].str.lower()
df
# SQL: UPDATE people SET EMAIL = LOWER(EMAIL);
```

```python
df['EMAIL'].apply(str.lower)     # preview only — result isn't assigned back yet
# SQL: SELECT LOWER(EMAIL) FROM people;
```

```python
def update_email(EMAIL):                             # MAKING CUSTOM FUNCTIONS
    return EMAIL.upper()

df['EMAIL'] = df['EMAIL'].apply(update_email)
df
# SQL: UPDATE people SET EMAIL = UPPER(EMAIL);
```

```python
df['EMAIL'] = df['EMAIL'].apply(lambda x: x.lower())   # USING LAMBDA FUNCTION
df
# SQL: UPDATE people SET EMAIL = LOWER(EMAIL);
```

```python
df.apply(len)   # CAN'T use apply directly per-element on a DataFrame — this runs len() per COLUMN,
                  # same as len(df['EMAIL']) but repeated for every column
# SQL: SELECT COUNT(first_name), COUNT(last_name), COUNT(EMAIL) FROM people;
```

```python
df.map(len)   # to apply a function to EVERY single cell of the DataFrame
# SQL: no direct equivalent — SQL has no built-in "apply this function to every cell in the table"
```

```python
df['first_name'] = df['first_name'].replace(
    {'Corey': 'Mihir', 'Jane': 'Purvesh', 'John': 'Nayana'}
)   # note: rename() was for COLUMN NAMES, replace() is for VALUES
df
# SQL: UPDATE people SET first_name = CASE first_name
#        WHEN 'Corey' THEN 'Mihir'
#        WHEN 'Jane'  THEN 'Purvesh'
#        WHEN 'John'  THEN 'Nayana'
#        ELSE first_name
#      END;
```

```python
df.apply(pd.Series.min)
# SQL: SELECT MIN(first_name), MIN(last_name), MIN(EMAIL) FROM people;
```
