# 01 · Pandas Basics — Series & DataFrame

SQL has **tables**. Pandas has two building blocks:

| Pandas | Roughly equivalent to |
|---|---|
| `Series` | a single column of a SQL table |
| `DataFrame` | a whole SQL table |

Every pandas line below has an SQL comment (`# SQL:`) showing the closest equivalent query.

```python
import pandas as pd

# A Series is a 1D labeled array — think of it as one column of query results.
s = pd.Series([10, 20, 30, 40])
print(s)
# SQL: SELECT * FROM (VALUES ROW(10),ROW(20),ROW(30),ROW(40)) AS t(val); 
```

```python
# You can give a Series custom labels (its "index") instead of default 0,1,2...
s = pd.Series([10, 20, 30], index=["a", "b", "c"])
print(s)
# CREATE TABLE vals (id TEXT PRIMARY KEY, val INT);
# INSERT INTO vals VALUES ('a',10), ('b',20), ('c',30);
```

```python
# A DataFrame is a 2D table with rows + columns, usually built from a dict of columns.
data = {
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 35],
    "city": ["Delhi", "Mumbai", "Bangalore"]
}
# SQL: CREATE TABLE people (name TEXT, age INT, city TEXT);
#      INSERT INTO people VALUES
#        ('Alice', 25, 'Delhi'),
#        ('Bob',   30, 'Mumbai'),
#        ('Charlie', 35, 'Bangalore');

df = pd.DataFrame(data)
print(df)
# SQL: SELECT * FROM people;

print(df[['name', 'age']])
# SQL: SELECT name, age FROM people;
```

```python
# A DataFrame can also be built from a list of rows, with column names given separately.
data = [
    ["Alice", 25],
    ["Bob", 30],
    ["Charlie", 35]
]

df = pd.DataFrame(data, columns=["Name", "Age"])
print(df)
# SQL: same result as:
# CREATE TABLE people2 (Name TEXT, Age INT);
# INSERT INTO people2 VALUES ('Alice',25), ('Bob',30), ('Charlie',35);
# SELECT * FROM people2;
```

```python
import numpy as np

arr = np.array([[1, 2], [3, 4]])
df = pd.DataFrame(arr, columns=["A", "B"])
print(df)
# SQL: CREATE TABLE t (A INT, B INT);
#      INSERT INTO t VALUES (1,2), (3,4);
#      SELECT * FROM t;
```
