# 03 · Selecting Data — `iloc`, `loc`, and `set_index`

- **`iloc`** = **i**nteger **loc**ation → select by numeric row/column *position*.
- **`loc`** = select by *label* (row/column name).

SQL tables are unordered sets of rows, so there's no true equivalent of "row position" —
the closest SQL gets is `LIMIT`/`OFFSET` after an `ORDER BY`.

```python
import os
import pandas as pd

os.chdir(r"C:\Users\purve\OneDrive\Desktop\extra\2019survey")
df = pd.read_csv('survey_results_public.csv')
schema_df = pd.read_csv('survey_results_schema.csv')
```

```python
df.iloc[1]   # the 2nd row (position 1)
# SQL: SELECT * FROM survey_results_public ORDER BY Respondent LIMIT 1 OFFSET 1;
```

```python
df.iloc[[0, 3]]   # 1st and 4th rows by position
# SQL: SELECT * FROM survey_results_public ORDER BY Respondent LIMIT 1 OFFSET 0
#      UNION ALL
#      SELECT * FROM survey_results_public ORDER BY Respondent LIMIT 1 OFFSET 3;
```

```python
df.iloc[[0, 3], 2]   # same two rows, but only the 3rd column (position 2)
# SQL: same as above but SELECT <3rd_column_name> instead of *
```

```python
df.loc[0:3, 'Hobbyist':'OpenSource']
# loc is LABEL based, and its slices are INCLUSIVE of the end label (unlike normal Python slicing)
# SQL: SELECT Hobbyist, ... , OpenSource
#      FROM survey_results_public
#      WHERE Respondent BETWEEN 0 AND 3;
```

```python
schema_df.iloc[[0], [1]]
# SQL: SELECT <2nd_column> FROM survey_results_schema ORDER BY Column LIMIT 1 OFFSET 0;
```

```python
schema_df.set_index('Column', inplace=True)
# Makes the 'Column' column the row label (index) instead of 0,1,2,...
# SQL: conceptually similar to: ALTER TABLE survey_results_schema ADD PRIMARY KEY (Column);
```

```python
schema_df.loc[['Hobbyist']]
# SQL: SELECT * FROM survey_results_schema WHERE Column = 'Hobbyist';
```

```python
schema_df.sort_index(ascending=False)
# not passing ascending= defaults to True → schema_df.sort_index()
# to keep the change permanent, add inplace=True
# SQL: SELECT * FROM survey_results_schema ORDER BY Column DESC;
```

### Setting an index on your own small DataFrame (`people` example)

```python
import pandas as pd

people_dict = {
    "first": ["Corey", "Jane", "John"],
    "last": ["Schafer", "Doe", "Doe"],
    "email": ["CoreyMSchafer@gmail.com", "JaneDoe@email.com", "JohnDoe@email.com"]
}
# SQL: CREATE TABLE people (first TEXT, last TEXT, email TEXT);
#      INSERT INTO people VALUES
#        ('Corey','Schafer','CoreyMSchafer@gmail.com'),
#        ('Jane','Doe','JaneDoe@email.com'),
#        ('John','Doe','JohnDoe@email.com');

# Use it to convert a Python variable you already have (dict/list/array) into a DataFrame.
df = pd.DataFrame(people_dict)
```

```python
df['email']
# SQL: SELECT email FROM people;
```

```python
df.email   # dot-notation shortcut, only works when the column name is a valid Python identifier
# SQL: SELECT email FROM people;
```

```python
df[['last', 'email']]
# SQL: SELECT last, email FROM people;
```

```python
df.set_index('email')   # PREVIEW only — not applied to df yet, since inplace isn't set
# SQL: (preview of) ALTER TABLE people ADD PRIMARY KEY (email);
```

```python
df.set_index('email', inplace=True)
df   # change is now permanent on df
# SQL: ALTER TABLE people ADD PRIMARY KEY (email);
```

```python
df.loc['CoreyMSchafer@gmail.com']
# SQL: SELECT * FROM people WHERE email = 'CoreyMSchafer@gmail.com';
```

```python
df.loc['CoreyMSchafer@gmail.com', 'last']
# SQL: SELECT last FROM people WHERE email = 'CoreyMSchafer@gmail.com';
```

```python
df.iloc[0]   # iloc still works after set_index, since it's purely position-based
# SQL: SELECT * FROM people LIMIT 1 OFFSET 0;   (position based, ignores the primary key)
```

```python
df.reset_index(inplace=True)   # back to the default 0,1,2,... index
# SQL: no exact equivalent — conceptually drops the custom key and goes back to a
# surrogate auto-increment id, e.g. ALTER TABLE people DROP PRIMARY KEY;
```
