```python
import pandas as pd

data = {
  'street': [
    '5230 Newell Road',
    '1381 9th Avenue',
    '1709 Broderick St',
    '2640 Steiner St',
    '2601 Lyon St',
    '1600 Amphitheatre Parkway',
    '7500 N Glenoaks Boulevard',
    '4501 Glencoe Avenue',
    '232 Santa Margarita Ave',
    '471 Emerson Street',
    '471 Emerson Street'
  ],
  'city': [
    'Palo Alto',
    'San Francisco',
    'San Francisco',
    'San Francisco',
    'San Francisco',
    'Mountain View',
    'Burbank',
    'Marina del Rey',
    'Menlo Park',
    None,
    None
  ],
  'state': [
    'CA',
    'CA',
    'CA',
    'CA',
    'CA',
    'CA',
    'CA',
    'CA',
    None,
    None,
    None
  ],
  'zip_code': [
    94303,
    94122,
    94115,
    94123,
    94123,
    94043,
    91504,
    90292,
    94025,
    None,
    None
  ],
  'delivery_type': [
    'First Class',
    'Standard',
    'Priority',
    'Standard',
    None,
    'Expresss',
    'Standard',
    'Standard',
    None,
    None,
    None
  ]
}

df = pd.DataFrame(data)
df
```

```python
cleaned_df = df.dropna()
cleaned_df
```

```python
# if we only care about missing values in specific columns
cleaned_df = df.dropna(subset = ['state', 'city'])
cleaned_df
```

```python
# The following code would fill in the missing values for the location column
df['delivery_type'] = df['delivery_type'].fillna('Standard')    #4
df
```

```python
# Finally, assuming the data is numerical, we can fill empty values with the average using .mean().
mean_zip_code = df['zip_code'].mean()

df['zip_code'] = df['zip_code'].fillna(mean_zip_code        )
```
